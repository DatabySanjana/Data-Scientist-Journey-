```python
# Create Book class 
class Book:
    def __init__(self, title, author, price):
        self.title = title
        self.author = author
        self.price = price
    def display(self):
        print("Title :", self.title)
        print("Author :", self.author)
        print("Price :", self.price)
book1 = Book("Python Programming", "Raju Mishra", 499)
book1.display()
```

```python
# Employee Class
class Employee:
    def __init__(self, name, salary, department):
        self.name = name
        self.salary = salary
        self.department = department
    def display(self):
        print("Name       :", self.name)
        print("Salary     :", self.salary)
        print("Department :", self.department)
    def bonus(self, amount):
        self.salary += amount
        print("Bonus Added Successfully!")
        print("New Salary :", self.salary)
employee1 = Employee("Rihan", 50000, "IT")
employee1.display()
print()
employee1.bonus(5000)
```

```python
# Bank Account 
class BankAccount:
    def __init__(self, account_number, balance):
        self.account_number = account_number
        self.balance = balance
    def deposit(self, amount):
        self.balance += amount
        print(amount, "deposited successfully.")
    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print(amount, "withdrawn successfully.")
        else:
            print("Insufficient Balance!")
    def check_balance(self):
        print("Current Balance :", self.balance)
account1 = BankAccount(123456789, 10000)
account1.check_balance()
account1.deposit(3000)
account1.check_balance()
account1.withdraw(5000)
account1.check_balance()
account1.withdraw(20000)
```

```python
# Create Movie class
class Movie:
    def __init__(self, title, rating, duration):
        self.title = title
        self.rating = rating
        self.duration = duration
    def play(self):
        print("Now Playing...")
        print("Title    :", self.title)
        print("Rating   :", self.rating)
        print("Duration :", self.duration, "minutes")
movie1 = Movie("Persuit of Happyness", 8.0, 117)
movie1.play()
```

```python
# Restuarant Billing System
class Restaurant:
    def __init__(self, customer_name, table_number):
        self.customer_name = customer_name
        self.table_number = table_number
        self.total_bill = 0
    def add_item(self, item, price):
        self.total_bill += price
        print(item, "added successfully.")
    def show_bill(self):
        print("\nCustomer Name :", self.customer_name)
        print("Table Number  :", self.table_number)
        print("Total Bill    :", self.total_bill)
    def pay_bill(self):
        print("\nPayment Successful!")
        print("Amount Paid :", self.total_bill)
        self.total_bill = 0
customer1 = Restaurant("Rahul", 5)
customer1.add_item("Pizza", 250)
customer1.add_item("Burger", 150)
customer1.add_item("Cold Drink", 50)
customer1.show_bill()
customer1.pay_bill()
```

```python
# Bus Reservation System
class Bus:
    def __init__(self, bus_number, route, seats):
        self.bus_number = bus_number
        self.route = route
        self.available_seats = seats
    def book_ticket(self):
        if self.available_seats > 0:
            self.available_seats -= 1
            print("Ticket Booked Successfully")
        else:
            print("No Seats Available")
    def show_details(self):
        print("\nBus Number :", self.bus_number)
        print("Route      :", self.route)
        print("Available Seats :", self.available_seats)
bus1 = Bus("WB_S20", "Kolkata to Digha", 40)
bus1.show_details()
bus1.book_ticket()
bus1.show_details()
```

```python
# Building College Database
class Student:
    def __init__(self, student_id, name, course, marks):
        self.student_id = student_id
        self.name = name
        self.course = course
        self.marks = marks
    def display(self):
        print("\nStudent ID :", self.student_id)
        print("Name       :", self.name)
        print("Course     :", self.course)
        print("Marks      :", self.marks)
    def grade(self):
        if self.marks >= 90:
            print("Grade : A")
        elif self.marks >= 80:
            print("Grade : B")
        elif self.marks >= 70:
            print("Grade : C")
        elif self.marks >= 60:
            print("Grade : D")
        else:
            print("Grade : F")
student1 = Student(1001, "Sanjana", "BSc", 86)
student1.display()
student1.grade()
```
