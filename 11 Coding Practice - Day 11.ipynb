### MEMORY MANAGEMENT

```python
# Python stores objects in memory
a = [10, 20, 30]
```

```python
# Variables are references
b = a
```

```python
b.append(40)
print(a)
print(b)
```

```python
# Let's prove it with id()
print(id(a))
print(id(b))
```

```python
# Shallow copy
p = [50, 60, 70]
q = p.copy()
q.append(40)
print(p)
print(q)
```

```python
# Deep Copy: If you want completely independent nested objects:
import copy
a = [[1, 2], [3, 4]]
b = copy.deepcopy(a)
b[0].append(99)
print(a)
print(b)
```

```python
# Reference Counting: Python keeps track of how many references point to an object
import sys
x = [1, 2, 3]
print(sys.getrefcount(x))
```

```python
y = x
print(sys.getrefcount(x))
```

```python
del y
print(sys.getrefcount(x))
```

```python
# del doesn't necessarily destroy the object
x = [1, 2, 3]
y = x
del x
print(y)
```

```python
# When does an object become unreachable
x = [1, 2, 3]
y = x
del x
del y
```

```python
print(x)
```

```python
print(y)
```

### MULTITHREADING
A thread is a unit of execution inside a process.

```python
import threading
def task():
    print("Task is running")
t1 = threading.Thread(target=task)
t2 = threading.Thread(target=task)
t1.start()
t2.start()
t1.join()
t2.join()
print("All tasks completed")
```

```python
# without threading
import time
def task(name):
    print(f"{name} started")
    time.sleep(2)
    print(f"{name} finished")
start = time.time()
task("Task 1")
task("Task 2")
task("Task 3")
end = time.time()
print("Time:", end - start)
```

```python
# with threading
import threading
import time
def task(name):
    print(f"{name} started")
    time.sleep(2)
    print(f"{name} finished")
start = time.time()
t1 = threading.Thread(target=task, args=("Task 1",))
t2 = threading.Thread(target=task, args=("Task 2",))
t3 = threading.Thread(target=task, args=("Task 3",))
t1.start()
t2.start()
t3.start()
t1.join()
t2.join()
t3.join()
end = time.time()
print("Time:", end - start)
```

### MULTIPROCESSING
A process is an independent running instance of a program.

```python
from multiprocessing import Process
def task():
    print("Task is running")
p1 = Process(target=task)
p1.start()
p1.join()
print("Task completed")
```

### ASYNCHRONOUS PROGRAMMING

```python
import asyncio
async def hello():
    print("Hello!")
await hello()
```

```python
await asyncio.sleep(2)
# It means: "I'm waiting here, so the event loop can handle other async work."
```

```python
import asyncio
async def task():
    print("Task started")
    await asyncio.sleep(2)
    print("Task finished")
await task()
```

```python
# With 2 tasks
import asyncio
async def task(name):
    print(name, "started")
    await asyncio.sleep(2)
    print(name, "finished")
async def main():
    await asyncio.gather(
        task("Task 1"),
        task("Task 2"))
await main()
```

```python
# asyncio.gather()
await asyncio.gather(
    task("Task 1"),
    task("Task 2"))
```

```python
import asyncio
async def task(name):
    print(name, "started")
    await asyncio.sleep(2)
    print(name, "finished")
async def main():
    await asyncio.gather(
        task("Task 1"),
        task("Task 2"))
await main()
```

```python
# Calling an async function doesn't immediately execute its body normally.
async def fetch():
    return 10
result = fetch()
```
