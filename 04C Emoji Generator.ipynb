#  Project: Feeling to Emoji Generator
Problem Statement: Write a Python program that acts as a Feeling to Emoji Generator using a dictionary.

The program should:

Store different feelings and their corresponding ASCII emojis in a dictionary.

Ask the user to enter how they are feeling.

Display the matching emoji if the feeling exists in the dictionary.

Display an appropriate message if the feeling is not found.

Allow the user to continue using the program until they choose to exit.

```python
# ASCII Emoji Translator
emojis = {
    ":)": "Smile",
    ":(": "Sad",
    ":D": "Laugh",
    ";)": "Wink",
    ":P": "Tongue Out",
    ":O": "Surprised"}
print("===== ASCII Emoji Translator =====")
emoji = input("Enter an ASCII Emoji: ")
if emoji in emojis:
    print("Meaning:", emojis[emoji])
else:
    print("Sorry! Emoji not found.")
```

```python
# Feeling to Emoji Generator
emojis = {
    "happy": ":)",
    "sad": ":(",
    "laugh": ":D",
    "wink": ";)",
    "surprised": ":O",
    "cool": "B)",
    "cry": ":'(",
    "love": "<3"}
print("===== Feeling to Emoji Generator =====")
feeling = input("How are you feeling today? ").lower()
if feeling in emojis:
    print("Emoji:", emojis[feeling])
else:
    print("Sorry! I don't know that feeling.")
```

```python
# Emoji Generator
emojis = {
    "happy": ":)",
    "sad": ":(",
    "laugh": ":D",
    "wink": ";)",
    "surprised": ":O",
    "cool": "B)",
    "love": "<3",
    "angry": ">:(",
    "sleepy": "-_-",
    "confused": ":/"}
while True:
    print("\n===== Feeling to Emoji Generator =====")
    print("Feelings Available:")
    print("happy, sad, laugh, wink, surprised")
    print("cool, love, angry, sleepy, confused")
    print("\nType 'exit' to quit.")
    feeling = input("\nEnter your feeling: ").lower()
    if feeling == "exit":
        print("Goodbye! (*^_^*) ")
        break
    elif feeling in emojis:
        print("Emoji Reaction:", emojis[feeling])
    else:
        print("Feeling not found!")
```
