# Student Performance Analyzer

A beginner NumPy project that analyzes marks of 100 students. It automatically calculates class statistics, grade distribution, top performers and demonstrates saving/loading NumPy arrays.

Features:

Generate marks for 100 students

Calculate mean, median, maximum, minimum and standard deviation

Display grade distribution (A, B, C, D, F)

Show Top 10 students

Save and load data using .npy files

Verify saved data using np.array_equal()

```python
# Generate marks for 100 students
import numpy as np
marks = np.random.randint(0, 101, 100)
print("Student Marks")
print(marks)
```

```python
print("========== Student Marks ==========")
print(marks)

# Class Statistis
print("\n========== Class Statistics ==========")
average = np.mean(marks)
median = np.median(marks)
highest = np.max(marks)
lowest = np.min(marks)
standard_deviation = np.std(marks)
print("Average Marks =", average)
print("Median Marks =", median)
print("Highest Marks =", highest)
print("Lowest Marks =", lowest)
print("Standard Deviation =", standard_deviation)
```

```python
# Grade Distribution
print("\n========== Grade Distribution ==========")
grade_a = np.sum((marks >= 90) & (marks <= 100))
grade_b = np.sum((marks >= 75) & (marks <= 89))
grade_c = np.sum((marks >= 60) & (marks <= 74))
grade_d = np.sum((marks >= 40) & (marks <= 59))
grade_f = np.sum(marks < 40)
print("A Grade Students =", grade_a)
print("B Grade Students =", grade_b)
print("C Grade Students =", grade_c)
print("D Grade Students =", grade_d)
print("F Grade Students =", grade_f)
```

```python
# Pass/Fail Statistics
pass_count = np.sum(marks >= 40)
fail_count = np.sum(marks < 40)
pass_percentage = (pass_count / len(marks)) * 100
fail_percentage = (fail_count / len(marks)) * 100
print("\n========== Pass / Fail ==========")
print("Pass Students =", pass_count)
print("Fail Students =", fail_count)
print("Pass Percentage = {:.2f}%".format(pass_percentage))
print("Fail Percentage = {:.2f}%".format(fail_percentage))
```

```python
# Top 10 students
print("\n========== Top 10 Students ==========")
ranking = np.argsort(marks)[::-1]
for i in range(10):
    student_id = ranking[i]
    print(f"Rank {i+1}: Student {student_id+1} - {marks[student_id]} Marks")
```

```python
# Save, Load & Verify
print("\n========== Save Student Marks ==========")
np.save("StudentMarks.npy", marks)
print("Marks Saved Successfully!")
print("\n========== Load Student Marks ==========")
loaded_marks = np.load("StudentMarks.npy")
print(loaded_marks)
print("\n========== Verification ==========")
if np.array_equal(marks, loaded_marks):
    print("Data Verified Successfully!")
else:
    print("Data Verification Failed!")
```
