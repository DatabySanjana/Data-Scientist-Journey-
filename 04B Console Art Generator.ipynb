## Console Art Generator
Start with: Square, Rectangle, Right Triangle, Pyramid, Inverted Pyramid, Diamond

Bonus: Hollow Square, Heart

```python
# Console Art Generator
while True:
    print("\n==============================")
    print("    CONSOLE ART GENERATOR")
    print("==============================")
    print("1. Square")
    print("2. Rectangle")
    print("3. Right Triangle")
    print("4. Pyramid")
    print("5. Inverted Pyramid")
    print("6. Diamond")
    print("7. Hollow Square")
    print("8. Heart")
    print("9. Exit")
    choice = input("\nEnter your choice: ")
# Square
    if choice == "1":
        size = int(input("Enter side length: "))
        print()
        for i in range(size):
            print("* " * size)
# Rectangle
    elif choice == "2":
        length = int(input("Enter length: "))
        width = int(input("Enter width: "))
        print()
        for i in range(width):
            print("* " * length)
# Right Triangle
    elif choice == "3":
        rows = int(input("Enter number of rows: "))
        print()
        for i in range(1, rows + 1):
            print("* " * i)
# Pyramid
    elif choice == "4":
        rows = int(input("Enter number of rows: "))
        print()
        for i in range(1, rows + 1):
            print(" " * (rows - i) + "* " * i)
# Inverted Pyramid
    elif choice == "5":
        rows = int(input("Enter number of rows: "))
        print()
        for i in range(rows, 0, -1):
            print(" " * (rows - i) + "* " * i)
# Diamond
    elif choice == "6":
        rows = int(input("Enter number of rows: "))
        print()
    # Upper Part
        for i in range(1, rows + 1):
            print(" " * (rows - i + 3) + "*" * (2 * i - 1))
    # Lower Part
        for i in range(rows - 1, 0, -1):
            print(" " * (rows - i + 3) + "*" * (2 * i - 1))
# Hollow Square
    elif choice == "7":
        size = int(input("Enter side length: "))
        print()
        for i in range(size):
            for j in range(size):
                if i == 0 or i == size - 1 or j == 0 or j == size - 1:
                    print("*", end=" ")
                else:
                    print(" ", end=" ")
            print()
# Heart
    elif choice == "8":
        print("""
          ***  ***
        ************
        ************
         **********
          ********
           ******
            ****
              *
""")
# Exit
    elif choice == "9":
        print("\nThank you for using Console Art Generator!")
        break
    else:
        print("\nInvalid Choice! Please try again.")
```
