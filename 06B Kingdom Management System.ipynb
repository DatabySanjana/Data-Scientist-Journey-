## Kingdom Management System
Version 1

Menu: Recruit Soldier, View Soldiers, Recruit Archer, Recruit Knight, Kingdom Information, Exit

Classes: Soldier, Archer, Knight, Kingdom

Concepts used: Class, Object, Constructor, Methods, Attributes, Lists of objectsKnight

```python
# Soldier Class
class Soldier:
    def __init__(self, name, health, attack):
        self.name = name
        self.health = health
        self.attack = attack
    def display(self):
        print("Name :", self.name)
        print("Health :", self.health)
        print("Attack :", self.attack)

soldier1 = Soldier("Jaswant Singh Rajput", 100, 25)
soldier1.display()
```

```python
# Archer class
class Archer:
    def __init__(self, name, health, arrows):
        self.name = name
        self.health = health
        self.arrows = arrows
    def display(self):
        print("Name :", self.name)
        print("Health :", self.health)
        print("Arrows :", self.arrows)

archer1 = Archer("Vikram Batra", 90, 40)
archer1.display()
```

```python
# Knight Class
class Knight:
    def __init__(self, name, health, armor):
        self.name = name
        self.health = health
        self.armor = armor
    def display(self):
        print("Name :", self.name)
        print("Health :", self.health)
        print("Armor :", self.armor)

knight1 = Knight("Arjun Singh", 150, 70)
knight1.display()
```

```python
# Kingdom Class - A class that stores everyone
# Recruit 
class Kingdom:
    def __init__(self, name):
        self.name = name
        self.soldiers = []
        self.archers = []
        self.knights = []
    def recruit_soldier(self):
        name = input("Enter Soldier Name : ")
        health = int(input("Enter Health : "))
        attack = int(input("Enter Attack : "))
        soldier = Soldier(name, health, attack)
        self.soldiers.append(soldier)

        print("Soldier Recruited Successfully!")
    def recruit_archer(self):
        name = input("Enter Archer Name : ")
        health = int(input("Enter Health : "))
        arrows = int(input("Enter Number of Arrows : "))
        archer = Archer(name, health, arrows)
        self.archers.append(archer)
        
        print("Archer Recruited Successfully!")
    def recruit_knight(self):
        name = input("Enter Knight Name : ")
        health = int(input("Enter Health : "))
        armor = int(input("Enter Armor : "))
        knight = Knight(name, health, armor)
        self.knights.append(knight)

        print("Knight Recruited Successfully!")
    def view_soldiers(self):
        if len(self.soldiers) == 0:
            print("No Soldiers Recruited.")
        else:
            print("\n--- Soldiers ---")
            for soldier in self.soldiers:
                soldier.display()
                print()

    def kingdom_information(self):
        print("\n===== Kingdom Information =====")
        print("Kingdom Name :", self.name)
        print("Soldiers :", len(self.soldiers))
        print("Archers  :", len(self.archers))
        print("Knights  :", len(self.knights))
        print("Total Army :", len(self.soldiers) + len(self.archers) + len(self.knights))
```

```python
# Main Program
kingdom = Kingdom("Vijayagarh")
while True:
    print("\n===== KINGDOM MANAGEMENT SYSTEM =====")
    print("1. Recruit Soldier")
    print("2. View Soldiers")
    print("3. Recruit Archer")
    print("4. Recruit Knight")
    print("5. Kingdom Information")
    print("6. Exit")
    choice = int(input("Enter your choice: "))
    if choice == 1:
        kingdom.recruit_soldier()
    elif choice == 2:
        kingdom.view_soldiers()
    elif choice == 3:
        kingdom.recruit_archer()
    elif choice == 4:
        kingdom.recruit_knight()
    elif choice == 5:
        kingdom.kingdom_information()
    elif choice == 6:
        print("Exiting Kingdom Management System...")
        break
    else:
        print("Invalid Choice! Please try again.")
```

## Kingdom Management System version2

Concepts to use: Inheritance, Polymorphism, Encapsulation, Abstraction

Classes: Character, Soldier, Archer, Knight, Wizard

Each character should have: Name, Health, Attack, Defence, Level

Methods: attack, train, show stats

Kingdom methods: recruit, view army, train army, kingdom info

```python
from abc import ABC, abstractmethod
# ---------------- ABSTRACT CLASS ----------------
class Character(ABC):
    def __init__(self, name, health, attack, defence, level):
        self.name = name                  
        self._health = health             
        self._attack = attack
        self._defence = defence
        self._level = level
    @abstractmethod
    def attack(self):
        pass
    def train(self):
        self._level += 1
        self._health += 20
        self._attack += 10
        self._defence += 5
        print(f"{self.name} trained successfully!")
    def show_stats(self):
        print(f"\nName     : {self.name}")
        print(f"Health   : {self._health}")
        print(f"Attack   : {self._attack}")
        print(f"Defence  : {self._defence}")
        print(f"Level    : {self._level}")
# ---------------- SOLDIER ----------------
class Soldier(Character):
    def attack(self):
        print(f"{self.name} attacks with a Sword!")
# ---------------- ARCHER ----------------
class Archer(Character):
    def attack(self):
        print(f"{self.name} shoots an Arrow!")
# ---------------- KNIGHT ----------------
class Knight(Character):
    def attack(self):
        print(f"{self.name} charges with a Lance!")
# ---------------- WIZARD ----------------
class Wizard(Character):
    def attack(self):
        print(f"{self.name} casts a Fire Spell!")
# ---------------- KINGDOM ----------------
class Kingdom:
    def __init__(self, name):
        self.name = name
        self.army = []
    def recruit(self):
        print("\nChoose Character")
        print("1. Soldier")
        print("2. Archer")
        print("3. Knight")
        print("4. Wizard")
        choice = int(input("Enter Choice : "))
        name = input("Enter Name : ")
        health = int(input("Health : "))
        attack = int(input("Attack : "))
        defence = int(input("Defence : "))
        level = int(input("Level : "))
        if choice == 1:
            character = Soldier(name, health, attack, defence, level)
        elif choice == 2:
            character = Archer(name, health, attack, defence, level)
        elif choice == 3:
            character = Knight(name, health, attack, defence, level)
        elif choice == 4:
            character = Wizard(name, health, attack, defence, level)
        else:
            print("Invalid Choice!")
            return
        self.army.append(character)
        print(f"{name} recruited successfully!")
    def view_army(self):
        if len(self.army) == 0:
            print("\nArmy is Empty!")
            return
        print("\n========== ARMY ==========")
        for character in self.army:
            character.show_stats()
            character.attack()
            print("-" * 30)
    def train_army(self):
        if len(self.army) == 0:
            print("No Army Available!")
            return
        for character in self.army:
            character.train()
        print("\nEntire Army Trained Successfully!")
    def kingdom_info(self):
        print("\n========== KINGDOM ==========")
        print("Kingdom :", self.name)
        print("Total Characters :", len(self.army))

        soldiers = 0
        archers = 0
        knights = 0
        wizards = 0
        for character in self.army:
            if isinstance(character, Soldier):
                soldiers += 1
            elif isinstance(character, Archer):
                archers += 1
            elif isinstance(character, Knight):
                knights += 1
            elif isinstance(character, Wizard):
                wizards += 1
        print("Soldiers :", soldiers)
        print("Archers  :", archers)
        print("Knights  :", knights)
        print("Wizards  :", wizards)
# ---------------- MAIN PROGRAM ----------------
kingdom = Kingdom("Vijayagarh")
while True:
    print("\n========== KINGDOM MANAGEMENT SYSTEM V2 ==========")
    print("1. Recruit Character")
    print("2. View Army")
    print("3. Train Army")
    print("4. Kingdom Information")
    print("5. Exit")
    choice = int(input("Enter Choice : "))
    if choice == 1:
        kingdom.recruit()
    elif choice == 2:
        kingdom.view_army()
    elif choice == 3:
        kingdom.train_army()
    elif choice == 4:
        kingdom.kingdom_info()
    elif choice == 5:
        print("Goodbye! ( *^ - ^* ")
        break
    else:
        print("Invalid Choice!")
```
