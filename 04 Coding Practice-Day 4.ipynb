### Lists

```python
#  Find the Largest element.
numbers = [12, 45, 7, 89, 34]
largest = max(numbers)
print("Largest Element =", largest)
```

```python
#  Find the Smallest element.
numbers = [12, 45, 7, 89, 34]
smallest = min(numbers)
print("Smallest Element =", smallest)
```

```python
# Find the  Second largest.
numbers = [12, 45, 7, 89, 34]
numbers.sort()
print("Second Largest =", numbers[-2])
```

```python
# Find the  Remove duplicates.
numbers = [10, 20, 10, 30, 20, 40]
unique = []
for i in numbers:
    if i not in unique:
        unique.append(i)
print(unique)
```

```python
# Find the  Reverse a list.
numbers = [10, 20, 30, 40, 50]
numbers.reverse()
print(numbers)
```

```python
# Find the  Sort ascending
numbers = [45, 12, 89, 7, 34]
numbers.sort()
print(numbers)
```

```python
#  Find the  Sort descending
numbers = [45, 12, 89, 7, 34]
numbers.sort(reverse=True)
print(numbers)
```

### Set

```python
# Find Common Elements
set1 = {10, 20, 30, 40}
set2 = {30, 40, 50, 60}
common = set1.intersection(set2)
print("Common Elements:", common)
```

```python
# Find Unique Elements
set1 = {10, 20, 30, 40}
set2 = {30, 40, 50, 60}
unique = set1.symmetric_difference(set2)
print("Unique Elements:", unique)
```

```python
# Remove Duplicates Using a Set
numbers = [10, 20, 10, 30, 20, 40, 50, 40]
unique = list(set(numbers))
print("Original List :", numbers)
print("Unique List   :", unique)
```

```python
# Union of two sets
set1 = {1, 2, 3}
set2 = {3, 4, 5}
result = set1.union(set2)
print("Union:", result)
```

```python
# Difference between two sets
set1 = {10, 20, 30, 40}
set2 = {30, 40, 50}
difference = set1.difference(set2)
print("Difference:", difference)
```

### Dictionaries

```python
# Search a key
student = {
    "Rahul": 90,
    "Priya": 95,
    "Amit": 80}
name = input("Enter Name: ")
if name in student:
    print(student[name])
else:
    print("Not Found")
```

```python
# Find the  Word frequency
sentence = input("Enter a sentence: ").lower()
words = sentence.split()
frequency = {}
for word in words:
    if word in frequency:
        frequency[word] += 1
    else:
        frequency[word] = 1
print("\nWord Frequency")
for word, count in frequency.items():
    print(word, ":", count)
```

```python
# Merge the dictionaries
dict1 = {
    "A": 10,
    "B": 20}
dict2 = {
    "C": 30,
    "D": 40}
dict1.update(dict2)
print(dict1)
```

```python
#  Find the Highest value
marks = {
    "Rahul": 85,
    "Priya": 92,
    "Amit": 78,
    "Neha": 95}
highest = " "
for student in marks:
    if highest == " " or marks[student] > marks[highest]:
        highest = student
print("Highest Scorer:", highest)
print("Marks:", marks[highest])
```

```python
#  Invert dictionary
student = {
    "A": 90,
    "B": 80,
    "C": 70}
inverse = {}
for key in student:
    inverse[student[key]] = key
print(inverse)
```

```python
#  Student marks dictionary
students = {}
n = int(input("How many students? "))
for i in range(n):
    name = input("Enter Name: ")
    marks = int(input("Enter Marks: "))
    students[name] = marks
print()
for name in students:
    print(name, ":", students[name])
```

```python
#  Average marks
students = {
    "Rahul": 85,
    "Priya": 90,
    "Amit": 75}
total = 0
for marks in students.values():
    total += marks
average = total / len(students)
print("Average =", average)
```

```python
#  Sort dictionary by value
students = {
    "Rahul": 85,
    "Priya": 92,
    "Amit": 78,
    "Neha": 95}
sorted_students = sorted(students.items(), key=lambda x: x[1])
for name, marks in sorted_students:
    print(name, ":", marks)
```
