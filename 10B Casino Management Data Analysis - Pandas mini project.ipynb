# Casino Management Data Analysis

```python
# Create the Dataset
import pandas as pd
import random

random.seed(42)
games = ["Poker", "Blackjack", "Roulette", "Slots", "Baccarat"]
cities = ["Mumbai", "Delhi", "Kolkata", "Bangalore", "Chennai"]
genders = ["Male", "Female"]
data = []
for i in range(1, 301):
    age = random.randint(18, 65)
    bet = random.randint(500, 10000)
    win = random.randint(0, 15000)
    data.append([
        i,
        f"Player_{i}",
        age,
        random.choice(genders),
        random.choice(games),
        bet,
        win,
        random.choice(cities),
        random.choice(["Yes", "No"])])
df = pd.DataFrame(data, columns=[
    "PlayerID",
    "Name",
    "Age",
    "Gender",
    "Game",
    "BetAmount",
    "WinAmount",
    "City",
    "VIP"])
df.to_csv("casino_data.csv", index=False)
print("Dataset Created Successfully!")
```

```python
# Feature Engineering
# Profit
df["Profit"] = df["WinAmount"] - df["BetAmount"]
```

```python
# ROI
df["ROI"] = df["WinAmount"] / df["BetAmount"]
```

```python
# Age Group
df["AgeGroup"] = df["Age"].apply(
    lambda x: "Young" if x < 30
    else "Adult" if x < 50
    else "Senior")
```

```python
# Big Bettor
df["BigBettor"] = df["BetAmount"] > 5000
```

```python
# High Winner
df["HighWinner"] = df["WinAmount"] > 10000
```

### Final Business Questions
1. The game with the highest average profit was ________.

2. The city with the highest total betting amount was ________.

3. The game with the highest average ROI was ________.

4. The highest-winning players were identified by comparing WinAmount.

5. The biggest losses were identified by comparing Profit.

```python
# 1. The game with the highest average profit was _______.
highest_avg_profit_game = df.groupby("Game")["Profit"].mean().idxmax()
print("The game with the highest average profit was:", highest_avg_profit_game)
```

```python
# 2. The city with the highest total betting amount was _______.
highest_betting_city = df.groupby("City")["BetAmount"].sum().idxmax()
print("The city with the highest total betting amount was:", highest_betting_city)
```

```python
# The game with the highest average ROI was ________.
highest_roi_game = df.groupby("Game")["ROI"].mean().idxmax()
print("The game with the highest average ROI was:", highest_roi_game)
```

```python
# 4. The highest-winning players were identified by comparing WinAmount.
top_winners = df.nlargest(10, "WinAmount")[["PlayerID", "Name", "Game", "WinAmount"]]
print(top_winners)
```

```python
# 5. The biggest losses were identified by comparing Profit.
biggest_losses = df.nsmallest(10, "Profit")[["PlayerID", "Name", "Game", "Profit"]]
print(biggest_losses)
```
