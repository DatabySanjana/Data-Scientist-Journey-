### Merge & Join

```python
# Create two DataFrames
df1 = pd.DataFrame({
    "PassengerId": [1, 2, 3, 4],
    "Name": ["Alice", "Bob", "Charlie", "David"]})
df2 = pd.DataFrame({
    "PassengerId": [1, 2, 3, 5],
    "Fare": [100, 200, 150, 300]})
```

```python
# Keeps only matching rows from both DataFrames
pd.merge(df1, df2, on="PassengerId")
```

```python
# Keeps all rows from the left DataFrame and matching rows from the right
pd.merge(df1, df2, on="PassengerId", how="left")
```

```python
# Keeps all rows from the right DataFrame and matching rows from the left
pd.merge(df1, df2, on="PassengerId", how="right")
```

```python
# Keeps all rows from both DataFrames, filling missing matches with NaN
pd.merge(df1, df2, on="PassengerId", how="outer")
```

```python
# Join
df1_index = df1.set_index("PassengerId")
df2_index = df2.set_index("PassengerId")
df1_index.join(df2_index)
```

```python
# Concat() - Stacks DataFrames together (by rows by default)
pd.concat([df1, df2])
```

### Window Functions

```python
# cumsum() - Running total
df["RunningFare"] = df["Fare"].cumsum()
df[["Fare", "RunningFare"]].head(10)
```

```python
# shift() - Moves values up or down
# The current row gets the previous row's fare
df["PreviousFare"] = df["Fare"].shift(1)
df[["Fare", "PreviousFare"]].head(10)
```

```python
# cummax() - Running maximum
df["MaxFare"] = df["Fare"].cummax()
df[["Fare", "MaxFare"]].head(10)
```

```python
# cummin() - Running minimum
df["MinFare"] = df["Fare"].cummin()
df[["Fare", "MinFare"]].head(10)
```

```python
# rank() - Ranks values
df["MinFare"] = df["Fare"].cummin()
df[["Fare", "MinFare"]].head(10)
```

```python
# rolling() - Moving average, Average of every 3 consecutive fares.
df["RollingFare"] = df["Fare"].rolling(window=3).mean()
df[["Fare", "RollingFare"]].head(10)
```

```python
# Create a rolling average of Age (window = 5)
df["Age_Rolling_Avg"] = df["Age"].rolling(window=5).mean()
print(df[["Age", "Age_Rolling_Avg"]])
```

### Time Series & Feature Creation

```python
# Create a sample DataFrame
dates = pd.DataFrame({"Date": ["2025-01-01", "2025-02-15", "2025-03-20"]})
dates
```

```python
# Convert String to Datetime format
dates["Date"] = pd.to_datetime(dates["Date"])
```

```python
# Creating new columns
dates["Year"] = dates["Date"].dt.year
dates["Month"] = dates["Date"].dt.month
dates["Day"] = dates["Date"].dt.day
dates["Weekday"] = dates["Date"].dt.day_name()
dates.head()
```

```python
# Create a new column showing the total family size (including the passenger)
df["FamilySize"] = df["SibSp"] + df["Parch"] + 1
df[["SibSp", "Parch", "FamilySize"]].head(10)
```

```python
# Create a Boolean column indicating whether the passenger is a child
df["IsChild"] = df["Age"] < 18
df[["Age", "IsChild"]].head(10)
```

```python
# Create a new column with each fare multiplied by 2
df["DoubleFare"] = df["Fare"].apply(lambda x: x * 2)
df[["Fare", "DoubleFare"]].head(10)
```

```python
# Categorize passengers into Child or Adult based on age
df["AgeGroup"] = df["Age"].apply(lambda x: "Child" if x < 18 else "Adult")
df[["Age", "AgeGroup"]].head(10)
```

```python
# Create a new column with fare increased by 10%
df["FareWithTax"] = df["Fare"] * 1.10
df[["Fare", "FareWithTax"]].head(10)
```
