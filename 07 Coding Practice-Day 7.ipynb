```python
## Reloading Modules using importlib.reload()

#When we import a module in Python, it is loaded into memory only once. 
#If we make changes to the module file after importing it, Python will continue using the old version stored in memory.

# To load the updated version of a module without restarting Python, we use importlib.reload().
```

```python
# Currency Calculator - Version 2
import importlib
import Currency

importlib.reload(Currency)

Amount = float(input("Enter amount in USD: "))
print("USD to INR =", Currency.usd_to_inr(Amount))
Amount2 = float(input("Enter amount in INR: "))
print("INR to USD =", Currency.inr_to_usd(Amount2))
Amount3 = float(input("Enter amount in Euro: "))
print("Euro to INR =", Currency.euro_to_inr(Amount3))
Amount4 = float(input("Enter amount in INR: "))
print("INR to Euro =", Currency.inr_to_euro(Amount4))
```

```python
# Scientific Calculator
import Scientific_Calculator as sc

print("Addition =", sc.add(10, 5))
print("Subtraction =", sc.subtract(10, 5))
print("Multiplication =", sc.multiply(10, 5))
print("Division =", sc.divide(10, 5))
print("Square =", sc.square(6))
print("Cube =", sc.cube(3))
print("Square Root =", sc.square_root(144))
print("Power =", sc.power(2, 5))
print("Factorial =", sc.factorial(5))
print("Absolute Value =", sc.absolute(-25))
print("Ceiling =", sc.ceiling(4.2))
print("Floor =", sc.floor(4.8))
print("Sin 90 =", sc.sine(90))
print("Cos 0 =", sc.cosine(0))
print("Tan 45 =", sc.tangent(45))
print("Natural Log =", sc.logarithm(10))
print("Log Base 10 =", sc.log10(100))
print("Value of Pi =", sc.pi_value())
print("Value of Euler's Number =", sc.e_value())
```
