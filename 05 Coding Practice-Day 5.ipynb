```python
# Merge two files
# Create File1
file = open("file1.txt", "w")
file.write("Hello\n")
file.write("Welcome to Python.")
file.close()
# Create File2
file = open("file2.txt", "w")
file.write("This is File 2.\n")
file.write("Have a nice day.")
file.close()
# Merge
file1 = open("file1.txt", "r")
file2 = open("file2.txt", "r")
data1 = file1.read()
data2 = file2.read()
file1.close()
file2.close()
merged = open("merged.txt", "w")
merged.write(data1)
merged.write("\n")
merged.write(data2)
merged.close()
print("Files merged successfully.")
```

```python
# Remove blank lines
file = open("student.txt", "r")
lines = file.readlines()
file.close()
print("Even Lines:")
for i in range(len(lines)):
    if (i + 1) % 2 == 0:
        print(lines[i].strip())
```

```python
# Read only even lines
file = open("student.txt", "r")
lines = file.readlines()
file.close()
print("Even Lines:")
for i in range(len(lines)):
    if (i + 1) % 2 == 0:
        print(lines[i].strip())
```

```python
# Read only odd lines
file = open("student.txt", "r")
lines = file.readlines()
file.close()
print("Odd Lines:")
for i in range(len(lines)):
    if (i + 1) % 2 != 0:
        print(lines[i].strip())
```

```python
# Find the longest line
file = open("student.txt", "r")
lines = file.readlines()
file.close()
longest = ""
for line in lines:
    if len(line) > len(longest):
        longest = line
print("Longest Line:")
print(longest)
```

```python
# Count vowels
file = open("student.txt", "r")
text = file.read()
file.close()
count = 0
for ch in text:
    if ch.lower() in "aeiou":
        count += 1
print("Total Vowels =", count)
```

```python
# Count uppercase and lowercase letters
file = open("student.txt", "r")
text = file.read()
file.close()
upper = 0
lower = 0
for ch in text:
    if ch.isupper():
        upper += 1
    elif ch.islower():
        lower += 1
print("Uppercase Letters =", upper)
print("Lowercase Letters =", lower)
```

## Exception Practice

```python
# Safe calculator
try:
    num1 = float(input("Enter First Number: "))
    num2 = float(input("Enter Second Number: "))
    op = input("Enter Operator (+,-,*,/): ")
    if op == "+":
        print("Answer =", num1 + num2)
    elif op == "-":
        print("Answer =", num1 - num2)
    elif op == "*":
        print("Answer =", num1 * num2)
    elif op == "/":
        print("Answer =", num1 / num2)
    else:
        print("Invalid Operator")
except:
    print("Invalid Input")
```

```python
# Handle divide by zero
try:
    a = int(input("Enter First Number: "))
    b = int(input("Enter Second Number: "))
    print("Answer =", a / b)
except ZeroDivisionError:
    print("Cannot Divide by Zero")
```

```python
# Invalid number input
try:
    num = int(input("Enter a Number: "))
    print("You entered", num)
except ValueError:
    print("Invalid Number")
```

```python
# File not found handler
try:
    file = open("abc.txt", "r")
    print(file.read())
    file.close()
except FileNotFoundError:
    print("File Not Found")
```

```python
# Index out of range
numbers = [10, 20, 30, 40, 50]
try:
    index = int(input("Enter Index: "))
    print(numbers[index])
except IndexError:
    print("Index Out of Range")
```

```python
# Dictionary key error
student = {
    "Rahul": 90,
    "Amit": 85,
    "Priya": 95}
try:
    name = input("Enter Student Name: ")
    print(student[name])
except KeyError:
    print("Student Not Found")
```

```python
# Type error example
try:
    print(10 + "Hello")
except TypeError:
    print("Type Error Occurred")
```

```python
# Value error example
try:
    age = int(input("Enter Age: "))
    print(age)
except ValueError:
    print("Please Enter Numbers Only")
```

```python
# Multiple exception handling
try:
    a = int(input("Enter First Number: "))
    b = int(input("Enter Second Number: "))
    print(a / b)
except ValueError:
    print("Invalid Number")
except ZeroDivisionError:
    print("Cannot Divide by Zero")
except TypeError:
    print("Type Error")
```

```python
# Nested try-except
try:
    try:
        a = int(input("Enter First Number: "))
        b = int(input("Enter Second Number: "))
        print(a / b)
    except ZeroDivisionError:
        print("Division by Zero")
except ValueError:
    print("Invalid Number")
```

```python
# Number validation
while True:
    try:
        num = int(input("Enter a Number: "))
        print("Valid Number")
        break
    except ValueError:
        print("Try Again")
```

```python
# Password validation
try:
    password = input("Enter Password: ")
    if len(password) < 8:
        raise Exception("Password Too Short")
    print("Password Accepted")
except Exception as e:
    print(e)
```

```python
# Safe file reader
try:
    name = input("Enter File Name: ")
    file = open(name, "r")
    print(file.read())
    file.close()
except FileNotFoundError:
    print("File Does Not Exist")
```

```python
# Raise a custom exception
age = int(input("Enter Age: "))
try:
    if age < 18:
        raise Exception("You are Not Eligible")
    print("Eligible")
except Exception as e:
    print(e)
```

```python
# Menu-driven program with exception handling
while True:
    print("\n1. Addition")
    print("2. Subtraction")
    print("3. Multiplication")
    print("4. Division")
    print("5. Exit")
    try:
        choice = int(input("Enter Choice: "))
        if choice == 5:
            print("Thank You")
            break
        a = float(input("Enter First Number: "))
        b = float(input("Enter Second Number: "))
        if choice == 1:
            print("Answer =", a + b)
        elif choice == 2:
            print("Answer =", a - b)
        elif choice == 3:
            print("Answer =", a * b)
        elif choice == 4:
            print("Answer =", a / b)
        else:
            print("Invalid Choice")
    except ZeroDivisionError:
        print("Cannot Divide by Zero")
    except ValueError:
        print("Invalid Input")
```
