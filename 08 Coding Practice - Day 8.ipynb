```python
# Merge two lists into one using iterators and print the final list
list1 = ["Apple", "Mango", "Orange"]
list2 = ["Banana", "Grapes", "Pineapple"]
iter1 = iter(list1)
iter2 = iter(list2)
merged = []
while True:
    try:
        merged.append(next(iter1))
    except StopIteration:
        break
while True:
    try:
        merged.append(next(iter2))
    except StopIteration:
        break
print("Merged List =", merged)
```

```python
# Find duplicate elements in a list using an iterator
numbers = [10, 20, 30, 20, 40, 10, 50, 30]
iterator = iter(numbers)
seen = []
duplicates = []
while True:
    try:
        value = next(iterator)
        if value in seen:
            if value not in duplicates:
                duplicates.append(value)
        else:
            seen.append(value)
    except StopIteration:
        break
print("Duplicate Elements:", duplicates)
```

```python
# Count the frequency of each word in a sentence using an iterator
sentence = input("Enter a sentence: ")
words = sentence.split()
iterator = iter(words)
frequency = {}
while True:
    try:
        word = next(iterator)
        if word in frequency:
            frequency[word] += 1
        else:
            frequency[word] = 1
    except StopIteration:
        break
for key, value in frequency.items():
    print(key, ":", value)
```

```python
# Check whether a list is sorted using an iterator
numbers = [10, 20, 30, 40, 50]
iterator = iter(numbers)
previous = next(iterator)
sorted_list = True
while True:
    try:
        current = next(iterator)
        if current < previous:
            sorted_list = False
            break
        previous = current
    except StopIteration:
        break
if sorted_list:
    print("List is Sorted")
else:
    print("List is Not Sorted")
```

```python
# Create your own custom iterator class that generates numbers from 1 to n
class Numbers:
    def __init__(self, n):
        self.n = n
        self.current = 1
    def __iter__(self):
        return self
    def __next__(self):
        if self.current <= self.n:
            number = self.current
            self.current += 1
            return number
        raise StopIteration
obj = Numbers(10)
for i in obj:
    print(i)
```

### Generator

```python
# Create a generator that yields numbers from 1 to 5
def numbers():
    for i in range(1, 6):
        yield i
gen = numbers()
for i in gen:
    print(i)
```

```python
# Create a generator that yields all the even numbers from 2 to 10
def even_numbers():
    for i in range(2, 11, 2):
        yield i
gen = even_numbers()
for i in gen:
    print(i)
```

```python
# Create a generator that yields each character of your name
def characters(name):
    for ch in name:
        yield ch
gen = characters("Sanjana")
for i in gen:
    print(i)
```

```python
# Create a generator that yields the squares of numbers from 1 to 10
def squares():
    for i in range(1, 11):
        yield i ** 2
gen = squares()
for i in gen:
    print(i)
```

```python
# Create a generator that yields each fruit from a list
def fruits():
    fruit_list = ["Apple", "Mango", "Orange", "Banana", "Grapes"]
    for fruit in fruit_list:
        yield fruit
gen = fruits()
for i in gen:
    print(i)
```

```python
# Create a generator that yields vowels from a string
def vowels(text):
    for ch in text:
        if ch.lower() in "aeiou":
            yield ch
gen = vowels("Programming")
for i in gen:
    print(i)
```

```python
# Create an infinite generator of natural numbers. Print only the first 15 numbers
def natural_numbers():
    n = 1
    while True:
        yield n
        n += 1
gen = natural_numbers()
for i in range(15):
    print(next(gen))
```

```python
# Create a Fibonacci generator
def fibonacci(n):
    a = 0
    b = 1
    for i in range(n):
        yield a
        a, b = b, a + b
gen = fibonacci(10)
for i in gen:
    print(i)
```

```python
# Create a generator that yields prime numbers between 1 and 100
def prime_numbers():
    for num in range(2, 101):
        prime = True
        for i in range(2, num):
            if num % i == 0:
                prime = False
                break
        if prime:
            yield num
for i in prime_numbers():
    print(i)
```

```python
# Create a generator that reads a list and yields only numbers divisible by 3 and 5
def divisible(numbers):
    for i in numbers:
        if i % 3 == 0 and i % 5 == 0:
            yield i
nums = [10, 15, 20, 30, 45, 60, 70, 75]
for i in divisible(nums):
    print(i)
```

### Decorator

```python
# Create a decorator that prints "Program Started" before a function runs
def start(func):
    def wrapper():
        print("Program Started")
        func()
    return wrapper
@start
def display():
    print("Hello World")
display()
```

```python
# Create a decorator that prints "Program Ended" after a function runs
def end(func):
    def wrapper():
        func()
        print("Program Ended")
    return wrapper
@end
def display():
    print("Hello World")
display()
```

```python
# Decorate a function that prints your name
def decorator(func):
    def wrapper():
        print("Before Function")
        func()
        print("After Function")
    return wrapper
@decorator
def name():
    print("Sanjana")
name()
```

```python
# Decorate a function that prints today's date
from datetime import date
def decorator(func):
    def wrapper():
        print("Today's Date:")
        func()
    return wrapper
@decorator
def today():
    print(date.today())
today()
```

```python
# Decorate a function that prints a welcome message
def welcome(func):
    def wrapper():
        print("Welcome User!")
        func()
    return wrapper
@welcome
def message():
    print("Hope you enjoy learning Python.")
message()
```

```python
# Create a decorator that counts how many times a function is called
count = 0
def counter(func):
    def wrapper():
        global count
        count += 1
        print("Function Called:", count, "times")
        func()
    return wrapper
@counter
def greet():
    print("Hello")
greet()
greet()
greet()
```

```python
# Create a decorator that prints the execution time of a function
import time
def timer(func):
    def wrapper():
        start = time.time()
        func()
        end = time.time()
        print("Execution Time:", end - start)
    return wrapper
@timer
def work():
    for i in range(1000000):
        pass
work()
```

```python
# Create a decorator that converts the returned string to uppercase
def uppercase(func):
    def wrapper():
        return func().upper()
    return wrapper
@uppercase
def message():
    return "hello python"
print(message())
```

```python
# Create a decorator that checks whether the user is logged in
logged_in = True
def login_required(func):
    def wrapper():
        if logged_in:
            func()
        else:
            print("Please Login First")
    return wrapper
@login_required
def dashboard():
    print("Welcome to Dashboard")
dashboard()
```

```python
# Create a decorator that logs the function name before execution
def logger(func):
    def wrapper():
        print("Running Function:", func.__name__)
        func()
    return wrapper
@logger
def display():
    print("Python Decorators")
display()
```

```python
# Create a decorator that accepts arguments
def greet(func):
    def wrapper(name):
        print("Welcome")
        func(name)
    return wrapper
@greet
def display(name):
    print("Hello", name)
display("Sanjana")
```

```python
# Create a decorator that works with any number of arguments (*args, **kwargs)
def decorator(func):
    def wrapper(*args, **kwargs):
        print("Before Function")
        return func(*args, **kwargs)
    return wrapper
@decorator
def add(a, b):
    print("Sum =", a + b)
add(10, 20)
```

```python
# Stack two decorators on one function
def first(func):
    def wrapper():
        print("First Decorator")
        func()
    return wrapper
def second(func):
    def wrapper():
        print("Second Decorator")
        func()
    return wrapper
@first
@second
def hello():
    print("Hello")
hello()
```

```python
# Create a decorator that repeats a function 3 times
def repeat(func):
    def wrapper():
        for i in range(3):
            func()
    return wrapper
@repeat
def hello():
    print("Welcome")
hello()
```

```python
# Create a decorator that catches exceptions and prints a custom error message
def exception_handler(func):
    def wrapper():
        try:
            func()
        except Exception as e:
            print("Something Went Wrong!")
            print(e)
    return wrapper
@exception_handler
def divide():
    a = 10
    b = 0
    print(a / b)
divide()
```

```python
# Create your own context manager using __enter__() and __exit__()
class MyContextManager:
    def __enter__(self):
        print("Entering Context Manager")
        return self
    def __exit__(self, exc_type, exc_value, traceback):
        print("Exiting Context Manager")
with MyContextManager():
    print("Inside the block")
```

```python
# Build a file logger using a custom context manager
class FileLogger:
    def __init__(self, filename):
        self.filename = filename
    def __enter__(self):
        self.file = open(self.filename, "a")
        return self.file
    def __exit__(self, exc_type, exc_value, traceback):
        self.file.close()
with FileLogger("log.txt") as logger:
    logger.write("Program Started\n")
    logger.write("User opened application\n")
    logger.write("Program Ended\n")
print("Log file created successfully!")
```

```python
# Measure how long a block of code takes using a context manager
import time
class Timer:
    def __enter__(self):
        self.start = time.time()
    def __exit__(self, exc_type, exc_value, traceback):
        self.end = time.time()
        print("Time Taken:",
              self.end - self.start,
              "seconds")
with Timer():
    for i in range(1000000):
        pass
```

```python
# Suppress a ZeroDivisionError using a custom context manager
class SuppressError:
    def __enter__(self):
        pass
    def __exit__(self, exc_type, exc_value, traceback):
        if exc_type == ZeroDivisionError:
            print("Zero Division Error Handled")
            return True
with SuppressError():
    print(10 / 0)
print("Program Continued")
```

```python
# Context manager that automatically opens and closes multiple files
# We will use our existing files: student.txt & student_copy.txt
class MultipleFiles:
    def __init__(self, file1, file2):
        self.file1_name = file1
        self.file2_name = file2
    def __enter__(self):
        self.file1 = open(self.file1_name, "r")
        self.file2 = open(self.file2_name, "r")
        return self.file1, self.file2
    def __exit__(self, exc_type, exc_value, traceback):
        self.file1.close()
        self.file2.close()
with MultipleFiles("student.txt", "student_copy.txt") as (file1, file2):
    print("Student File:")
    print(file1.read())
    print("Copied File:")
    print(file2.read())
print("Both files closed successfully!")
```
